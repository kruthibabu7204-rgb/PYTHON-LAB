```python
#program to find the largest elements in a list
numbers=[12,45,35,89,34]
largest=max(numbers)
print("list:",numbers)
print("largest element:",largest)
```

```python
#program to find the smallest elements in a list
numbers=[12,45,35,89,34]
largest=min(numbers)
print("list:",numbers)
print("largest element:",largest)
```
