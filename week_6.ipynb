```python
##function with parameters
def add(a,b):
    return a + b
print(add(3,5))
```

```python
##functions from default argument
def power(base, exp=2):
    return base ** exp
print(power(4))  #defult squares
print(power(4, 3))
```

```python
##function returning multiple values
def stats(numbers):
    return min(numbers), max(numbers), sum(numbers)/len(numbers)
    print(stats([2, 4, 6, 8]))
```
