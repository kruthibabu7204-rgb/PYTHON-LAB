```python
##palindrome with recursion
def palindrome(word):
    if len(word)<=1:
        return True
    if word[0]==word[-1]:
       return palindrome(word[1:-1])
    else:
       return False
text=input("enter a ")
if(palindrome(text)):
    print("palindrome")
else:
    print("not a palindrome")
```

```python
def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    
    return is_palindrome(s[1:-1])

word = "madam"
if is_palindrome(word):
    print(f"'{word}' is a palindrome")
else:
    print(f"'{word}' is not a palindrome")
```
